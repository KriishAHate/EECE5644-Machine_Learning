### Logan airport NOAA

```python
# import libraries
import pandas as pd
import numpy as np

# list of input files (add more file if required)
files = [
    "NOAA data/NOAA Logan Weather Filtered Data Jan-Nov 2025.csv",
]

# read and stack all the years together
df_list = []

for file in files:
    df_raw = pd.read_csv(file)
    print(f"Loaded {file} with shape {df_raw.shape}")
    df_list.append(df_raw)

# concatenate
df_raw = pd.concat(df_list, ignore_index=True)

display(df_raw[2000:3000])
print(df_raw.dtypes)
df_raw.shape
```
